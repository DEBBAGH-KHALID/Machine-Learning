```python
# importation des packages
import numpy as np
import pandas as pd
from sklearn import datasets
import tensorflow as tf
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
# recuperation du dataset
df = datasets.load_breast_cancer()
# description sur dataset
print(df.DESCR)
```

```python
# les données d'entrée
X = df.data
```

```python
# les données de sortie
Y = df.target
```

```python
# standardiser les donnees
scaler = StandardScaler()
x_scaled = scaler.fit_transform(X)
x_scaled
```

```python
# diviser data
X_train, X_test, y_train, y_test = train_test_split(x_scaled, Y, test_size=0.2, random_state=30)
```

```python
# Creation du reseau de neurone
model = tf.keras.models.Sequential()
# couche cachee avec d'entree
model.add(tf.keras.layers.Dense(units=15,activation='relu', kernel_initializer='uniform',input_dim=30))
```

```python
# couche de sortie
model.add(tf.keras.layers.Dense(units=1,activation='sigmoid', kernel_initializer='uniform'))
```

```python
# optimizer='adam' : comment ajuster les poids synaptiques
# loss='binary_crossentropy' : calcule la fonction cout
# metrics=['binary_accuracy'] : calcule l'accuracy
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['binary_accuracy'])
```

```python
model.summary()
```

```python
# batch_size=10 : les donnees diviser sur des morceaux de 10 lignes donc 569 / 10 = 57, alors pour une epoch modele fait 57 mise a jour de poids.

model.fit(X_train,y_train,batch_size=10,epochs=100)
```

```python
# tester le modele
test_loss,test_accuracy=model.evaluate(X_test,y_test)
print("test accuracy",test_accuracy)
print("test loss",test_loss)
```
